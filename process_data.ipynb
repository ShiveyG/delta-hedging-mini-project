```python
import pandas as pd

# choose strike price
k = 25300

# load data
opt = pd.read_csv(r"Options Data\NIFTY_27-Mar-2025\OPTIDX_NIFTY_CE_01-Mar-2025_TO_31-Mar-2025.csv")
fut = pd.read_csv(r"Options Data\NIFTY_27-Mar-2025\FUTIDX_NIFTY_01-Mar-2025_TO_31-Mar-2025 (1).csv")

opt.columns = opt.columns.str.strip()
fut.columns = fut.columns.str.strip()

# print(opt.head())
# print(fut.head())
# convert dates
opt["Date"] = pd.to_datetime(opt["Date"])
fut["Date"] = pd.to_datetime(fut["Date"])

opt["Expiry"] = pd.to_datetime(opt["Expiry"])
fut["Expiry"] = pd.to_datetime(fut["Expiry"])

# print(opt["Strike Price"].unique())

# filter CE options with strike k
opt_filtered = opt[
    (opt["Option type"] == "CE") &
    (opt["Strike Price"] == k)
]
# print(opt_filtered.head())
merged = pd.merge(
    opt_filtered,
    fut,
    on=["Symbol", "Date", "Expiry"],
    suffixes=("_OPT", "_FUT")
)

merged["Time to Maturity"] = (merged["Expiry"] - merged["Date"]).dt.days / 365
# final dataframe with required columns
output = pd.DataFrame({
    "Date": merged["Date"],
    "Spot Price": merged["Underlying Value_OPT"],
    "Futures Price": merged["Close_FUT"],
    "Option Market Price": merged["Close_OPT"],
    "Strike Price": merged["Strike Price"],
    "Expiry Date": merged["Expiry"],
    "Time to Maturity": merged["Time to Maturity"]
})

print(output.head())
output.to_csv(r"output_otm.csv", index=False)
```
